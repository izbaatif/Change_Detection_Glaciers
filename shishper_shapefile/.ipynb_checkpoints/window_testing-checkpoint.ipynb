```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.metrics import structural_similarity as ssim
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/april2_shishper.tif"
img_end_file   = "shishper_aligned/april17_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"

sar_sig_files = [
    "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"
]

window_size = 30
radius = window_size // 2
min_dist = 2 * radius
band_nir = 4
lat_cutoff = 36.350

# ---------------------------
# 2. HELPER FUNCTION (SSIM ONLY)
# ---------------------------
def get_masked_ssim(img1, img2, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    if w1.shape[0] < 7 or w1.shape[1] < 7:
        return None

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    mask = x**2 + y**2 <= rad**2

    data_rng = float(w2.max() - w2.min()) if w2.max() != w2.min() else 1.0
    _, ssim_map = ssim(w1, w2, data_range=data_rng, win_size=7, full=True)

    return ssim_map[mask].mean()

# ---------------------------
# 3. DATA LOADING & MASKING
# ---------------------------
print("Loading images and shapefile...")

with rasterio.open(img_start_file) as ds:
    img_start = ds.read().astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read().astype(np.float32)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# ✅ FIXED latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. MAP SAR POINTS TO PLANETSCOPE
# ---------------------------
sig_rows_ps, sig_cols_ps = [], []

for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)

        xs, ys = rasterio.transform.xy(
            ds_sar.transform, r_sar, c_sar, offset="center"
        )

        cols_p, rows_p = ~transform_ps * (xs, ys)

        # ✅ FIXED rounding
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_rows_ps.append(r)
                    sig_cols_ps.append(c)

sig_coords = list(set(zip(sig_rows_ps, sig_cols_ps)))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)

print(f"Verified {num_pts} significant points inside glacier boundary.")

# ---------------------------
# 5. RANDOM CONTROL POINTS
# ---------------------------
rand_coords = []
attempts = 0

print(f"Sampling {num_pts} random control points on glacier...")

while len(rand_coords) < num_pts and attempts < 20000:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)

    if glacier_mask[r, c]:
        if sig_coords_arr.size > 0:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.any(dist_to_sig < min_dist):
                attempts += 1
                continue

        if rand_coords:
            dist_to_rand = np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1)
            if np.any(dist_to_rand < min_dist):
                attempts += 1
                continue

        rand_coords.append((r, c))

    attempts += 1

# ---------------------------
# 6. CALCULATE SSIM
# ---------------------------
sig_ssim = [
    get_masked_ssim(img_start[band_nir], img_end[band_nir], r, c, radius)
    for r, c in sig_coords
]
rand_ssim = [
    get_masked_ssim(img_start[band_nir], img_end[band_nir], r, c, radius)
    for r, c in rand_coords
]

sig_ssim = [v for v in sig_ssim if v is not None]
rand_ssim = [v for v in rand_ssim if v is not None]

# ---------------------------
# 7. STATS & VISUALIZATION
# ---------------------------
u, p = mannwhitneyu(sig_ssim, rand_ssim, alternative="less")

print("\n" + "=" * 45)
print(f"RESULTS: Significant (N={len(sig_ssim)}) vs Random (N={len(rand_ssim)})")
print("=" * 45)
print(f"Avg SSIM - Sig: {np.mean(sig_ssim):.4f}")
print(f"Avg SSIM - Rand: {np.mean(rand_ssim):.4f}")
print(f"Mann–Whitney U p-value: {p:.6f}")
print("=" * 45)

# ---------------------------
# 8. PLOTTING
# ---------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start[band_nir], (2, 98))
ax1.imshow(img_start[band_nir], cmap="gray", vmin=vmin, vmax=vmax)

gdf.boundary.plot(ax=ax1, color="lime", linewidth=1, label="Glacier Boundary")
ax1.axhline(row_limit, color="yellow", linestyle="--", label="Lat Cutoff")

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, fill=False, edgecolor="red", linewidth=0.8))

for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, fill=False, edgecolor="cyan", linewidth=0.8))

ax1.set_title("Window Locations (Red = SAR, Cyan = Random)")
ax1.legend()

ax2.boxplot([sig_ssim, rand_ssim], labels=["Significant", "Random"])
ax2.set_ylabel("SSIM")
ax2.set_title("SSIM Comparison (Lower = More Change)")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/aug12_shishper.tif"
img_end_file   = "shishper_aligned/aug26_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"

sar_sig_files = [
    "max_ratios_significant/ratio_significant_99_2025-08-03.tif",
    "max_ratios_significant/ratio_significant_99_2025-08-15.tif",
    "max_ratios_significant/ratio_significant_99_2025-08-27.tif"
]

window_size = 75
radius = window_size // 2
min_dist = 2 * radius 
band_nir = 4           
lat_cutoff = 36.350   

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1, w2 = img1[r0:r1, c0:c1], img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    mask = x**2 + y**2 <= rad**2
    return gms_map[mask].mean()

# ---------------------------
# 3. DATA LOADING & MASKING
# ---------------------------
with rasterio.open(img_start_file) as ds:
    img_start = ds.read(band_nir).astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read(band_nir).astype(np.float32)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# ✅ FIXED latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. MAP SAR POINTS TO PLANETSCOPE
# ---------------------------
sig_coords = []
for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)

        # ✅ FIXED snapping
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)
print(f"Verified {num_pts} significant points.")

# ---------------------------
# 5. SPATIAL REJECTION SAMPLING
# ---------------------------
rand_coords = []
attempts = 0
while len(rand_coords) < num_pts and attempts < 20000:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)
    if glacier_mask[r, c]:
        dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
        if np.all(dist_to_sig >= min_dist):
            if not rand_coords or np.all(
                np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist
            ):
                rand_coords.append((r, c))
    attempts += 1

# ---------------------------
# 6. CALCULATE FSIM (GMS)
# ---------------------------
sig_fsim = [get_masked_fsim_metrics(img_start, img_end, r, c, radius) for r, c in sig_coords]
rand_fsim = [get_masked_fsim_metrics(img_start, img_end, r, c, radius) for r, c in rand_coords]

sig_fsim = [f for f in sig_fsim if f is not None]
rand_fsim = [f for f in rand_fsim if f is not None]

# ---------------------------
# 7. STATS & VISUALIZATION
# ---------------------------
u_fsim, p_fsim = mannwhitneyu(sig_fsim, rand_fsim, alternative='less')

print("\n" + "="*45)
print("RESULTS: FEATURE SIMILARITY (FSIM / GMS)")
print("="*45)
print(f"Avg FSIM - Sig: {np.mean(sig_fsim):.4f} | Rand: {np.mean(rand_fsim):.4f}")
print(f"P-value: {p_fsim:.6f}")
print("="*45)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start, (2, 98))
ax1.imshow(img_start, cmap='gray', vmin=vmin, vmax=vmax)

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='red', fill=False, linewidth=0.8))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='cyan', fill=False, linewidth=0.8))

ax1.set_title("Feature Analysis Windows")

ax2.boxplot([sig_fsim, rand_fsim], labels=['Significant', 'Random'])
ax2.set_title("FSIM (GMS) — Lower = Greater Structural Change")
ax2.set_ylabel("Similarity Score")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/july5_shishper.tif"
img_end_file   = "shishper_aligned/july21_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"  # Update this

sar_sig_files = [
    "max_ratios_significant/ratio_significant_99_2025-07-10.tif",
    #"max_ratios_significant/ratio_significant_99_2025-08-15.tif",
    "max_ratios_significant/ratio_significant_99_2025-07-22.tif"
]


window_size = 70
radius = window_size // 2
min_dist = 2 * radius 
band_nir = 4           
lat_cutoff = 36.350   

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1, w2 = img1[r0:r1, c0:c1], img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    mask = x**2 + y**2 <= rad**2
    return gms_map[mask].mean()

# ---------------------------
# 3. DATA LOADING & MASKING
# ---------------------------
with rasterio.open(img_start_file) as ds:
    img_start = ds.read(band_nir).astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read(band_nir).astype(np.float32)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# ✅ FIXED latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. MAP SAR POINTS TO PLANETSCOPE
# ---------------------------
sig_coords = []
for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)

        # ✅ FIXED snapping
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)
print(f"Verified {num_pts} significant points.")

# ---------------------------
# 5. SPATIAL REJECTION SAMPLING
# ---------------------------
rand_coords = []
attempts = 0
while len(rand_coords) < num_pts and attempts < 20000:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)
    if glacier_mask[r, c]:
        dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
        if np.all(dist_to_sig >= min_dist):
            if not rand_coords or np.all(
                np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist
            ):
                rand_coords.append((r, c))
    attempts += 1

# ---------------------------
# 6. CALCULATE FSIM (GMS)
# ---------------------------
sig_fsim = [get_masked_fsim_metrics(img_start, img_end, r, c, radius) for r, c in sig_coords]
rand_fsim = [get_masked_fsim_metrics(img_start, img_end, r, c, radius) for r, c in rand_coords]

sig_fsim = [f for f in sig_fsim if f is not None]
rand_fsim = [f for f in rand_fsim if f is not None]

# ---------------------------
# 7. STATS & VISUALIZATION
# ---------------------------
u_fsim, p_fsim = mannwhitneyu(sig_fsim, rand_fsim, alternative='less')

print("\n" + "="*45)
print("RESULTS: FEATURE SIMILARITY (FSIM / GMS)")
print("="*45)
print(f"Avg FSIM - Sig: {np.mean(sig_fsim):.4f} | Rand: {np.mean(rand_fsim):.4f}")
print(f"P-value: {p_fsim:.6f}")
print("="*45)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start, (2, 98))
ax1.imshow(img_start, cmap='gray', vmin=vmin, vmax=vmax)

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='red', fill=False, linewidth=0.8))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='cyan', fill=False, linewidth=0.8))

ax1.set_title("Feature Analysis Windows")

ax2.boxplot([sig_fsim, rand_fsim], labels=['Significant', 'Random'])
ax2.set_title("FSIM (GMS) — Lower = Greater Structural Change")
ax2.set_ylabel("Similarity Score")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/aug12_shishper.tif"
img_end_file   = "shishper_aligned/aug26_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp" 

sar_sig_files = [
    "max_ratios_significant/ratio_significant_99_2025-08-03.tif",
    "max_ratios_significant/ratio_significant_99_2025-08-15.tif",
    "max_ratios_significant/ratio_significant_99_2025-08-27.tif"
]

window_size = 70
radius = window_size // 2
min_dist = 2 * radius 

# PlanetScope Band Mapping (approximate for SuperDove)
# NDSI Proxy: (Green - NIR) / (Green + NIR) 
band_green = 2 
band_nir = 4   
lat_cutoff = 36.350   

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_ndsi(green_band, nir_band):
    """Calculates NDSI proxy. Adding epsilon to avoid divide by zero."""
    num = (green_band - nir_band)
    den = (green_band + nir_band)
    # Mask zeros in denominator
    den[den == 0] = 1e-9
    return num / den

def get_ndsi_change_metrics(img1_stack, img2_stack, r, c, rad):
    """Calculates the absolute change in NDSI within a circular radius."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    
    if r0 < 0 or r1 > img1_stack.shape[1] or c0 < 0 or c1 > img1_stack.shape[2]:
        return None

    # Extract windows for Green and NIR
    g1 = img1_stack[band_green-1, r0:r1, c0:c1]
    n1 = img1_stack[band_nir-1, r0:r1, c0:c1]
    g2 = img2_stack[band_green-1, r0:r1, c0:c1]
    n2 = img2_stack[band_nir-1, r0:r1, c0:c1]
    
    # Calculate NDSI for both dates
    ndsi1 = calculate_ndsi(g1, n1)
    ndsi2 = calculate_ndsi(g2, n2)
    
    # Create circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    mask = x**2 + y**2 <= rad**2
    
    # Calculate Absolute Change in NDSI
    delta_ndsi = np.abs(ndsi2 - ndsi1)
    mean_change = delta_ndsi[mask].mean()
    
    return mean_change

# ---------------------------
# 3. DATA LOADING & MASKING
# ---------------------------
print("Loading images and shapefile...")
with rasterio.open(img_start_file) as ds:
    img_start = ds.read().astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read().astype(np.float32)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

row_limit = int((lat_cutoff - transform_ps.f) / transform_ps.e)
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. MAP SAR POINTS TO PLANETSCOPE
# ---------------------------
sig_coords = []
for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        sar_data = ds_sar.read(1)
        r_sar, c_sar = np.where(sar_data > 0)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar)
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p, cols_p = np.array(rows_p).astype(int), np.array(cols_p).astype(int)
        
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)
print(f"Found {num_pts} significant points.")

# ---------------------------
# 5. SPATIAL REJECTION SAMPLING
# ---------------------------
rand_coords = []
attempts = 0
while len(rand_coords) < num_pts and attempts < 20000:
    r, c = random.randint(radius, row_limit - radius - 1), random.randint(radius, W - radius - 1)
    if glacier_mask[r, c]:
        dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
        if np.all(dist_to_sig >= min_dist):
            if not rand_coords or np.all(np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist):
                rand_coords.append((r, c))
    attempts += 1

# ---------------------------
# 6. CALCULATE NDSI METRICS
# ---------------------------
sig_ndsi_change = [get_ndsi_change_metrics(img_start, img_end, r, c, radius) for r, c in sig_coords]
rand_ndsi_change = [get_ndsi_change_metrics(img_start, img_end, r, c, radius) for r, c in rand_coords]

# Filter Nones
sig_ndsi_change = [v for v in sig_ndsi_change if v is not None]
rand_ndsi_change = [v for v in rand_ndsi_change if v is not None]

# ---------------------------
# 7. STATS & VISUALIZATION
# ---------------------------
# Alternative 'greater' because we expect more NDSI change at significant points
u_stat, p_val = mannwhitneyu(sig_ndsi_change, rand_ndsi_change, alternative='greater')

print("\n" + "="*45)
print(f"NDSI CHANGE RESULTS")
print("="*45)
print(f"Avg ΔNDSI (Sig): {np.mean(sig_ndsi_change):.4f}")
print(f"Avg ΔNDSI (Rand): {np.mean(rand_ndsi_change):.4f}")
print(f"P-value: {p_val:.6f}")
print("="*45)

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Show NIR band as background
ax1.imshow(img_start[band_nir-1], cmap='gray', vmin=np.percentile(img_start[band_nir-1], 5), vmax=np.percentile(img_start[band_nir-1], 95))
for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='red', fill=False))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='cyan', fill=False))
ax1.set_title("Analysis Windows (Red=Hypothesis, Cyan=Control)")

# Boxplot
ax2.boxplot([sig_ndsi_change, rand_ndsi_change], labels=['Significant Points', 'Random Points'])
ax2.set_title("Magnitude of NDSI Change")
ax2.set_ylabel("Abs(ΔNDSI)")

plt.show()
```

```python
# April 5th — Glacier-masked FSIM analysis
# Random points may overlap each other, but not significant points

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/april2_shishper.tif"
img_end_file   = "shishper_aligned/april17_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"

sar_sig_files = [
    "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"
]

vv_mask_file = "bursts/2025_04_05/VV_shape.tif"
vh_mask_file = "bursts/2025_04_05/VH_shape.tif"

window_size = 25
radius = window_size // 2
min_dist = 2 * radius
band_nir = 8
lat_cutoff = 36.350

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.count_nonzero(final_mask) == 0:
        return None

    return gms[final_mask].mean()

def reproject_to_ref(src_path, ref_shape, ref_transform, ref_crs):
    with rasterio.open(src_path) as src:
        data = src.read(1)
        dst = np.zeros(ref_shape, dtype=np.float32)

        reproject(
            source=data,
            destination=dst,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=ref_transform,
            dst_crs=ref_crs,
            dst_height=ref_shape[0],
            dst_width=ref_shape[1],
            resampling=Resampling.nearest
        )
    return dst


# ---------------------------
# 3. LOAD DATA & MASKS
# ---------------------------
with rasterio.open(img_start_file) as ds:
    img_start = ds.read(band_nir).astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read(band_nir).astype(np.float32)

from rasterio.warp import reproject, Resampling

# Reproject VV and VH to PlanetScope grid
vv = reproject_to_ref(
    "bursts/2025_04_05/VV_shape.tif",
    (H, W),
    transform_ps,
    crs_ps
)

vh = reproject_to_ref(
    "bursts/2025_04_05/VH_shape.tif",
    (H, W),
    transform_ps,
    crs_ps
)


gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

sar_valid_mask = (vv > 0) | (vh > 0)
valid_mask = glacier_mask & sar_valid_mask

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. SIGNIFICANT POINTS
# ---------------------------
sig_coords = []

for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)

        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)

print(f"Significant points: {num_pts}")

# ---------------------------
# 5. RANDOM POINTS (CAN OVERLAP EACH OTHER)
# ---------------------------
rand_coords = []
attempts = 0
max_attempts = 30000

while len(rand_coords) < num_pts and attempts < max_attempts:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)

    if not valid_mask[r, c]:
        attempts += 1
        continue

    candidate = np.array([r, c])

    # ONLY enforce distance from significant points
    if np.any(np.linalg.norm(sig_coords_arr - candidate, axis=1) < min_dist):
        attempts += 1
        continue

    rand_coords.append((r, c))
    attempts += 1

print(f"Random points accepted: {len(rand_coords)}")

# ---------------------------
# 6. FSIM
# ---------------------------
sig_fsim = [
    get_masked_fsim_metrics(img_start, img_end, valid_mask, r, c, radius)
    for r, c in sig_coords
]

rand_fsim = [
    get_masked_fsim_metrics(img_start, img_end, valid_mask, r, c, radius)
    for r, c in rand_coords
]

sig_fsim = [v for v in sig_fsim if v is not None]
rand_fsim = [v for v in rand_fsim if v is not None]

# ---------------------------
# 7. STATS & PLOTS
# ---------------------------
u, p = mannwhitneyu(sig_fsim, rand_fsim, alternative="less")

print("\n" + "="*50)
print("FSIM RESULTS (Randoms may overlap)")
print("="*50)
print(f"Mean FSIM | Sig   : {np.mean(sig_fsim):.4f}")
print(f"Mean FSIM | Rand  : {np.mean(rand_fsim):.4f}")
print(f"P-value   : {p:.6e}")
print("="*50)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start, (2, 98))
ax1.imshow(img_start, cmap="gray", vmin=vmin, vmax=vmax)

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, edgecolor="red", fill=False, lw=0.8))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, edgecolor="cyan", fill=False, lw=0.8))

ax1.set_title("FSIM Windows (Randoms can overlap)")

ax2.boxplot([sig_fsim, rand_fsim], labels=["Significant", "Random"])
ax2.set_ylabel("FSIM (GMS)")
ax2.set_title("Lower = More Structural Change")

plt.tight_layout()
plt.show()
```

```python
# April 17th — Glacier-masked FSIM analysis
# Random points may overlap each other, but not significant points

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/april17_shishper.tif"
img_end_file   = "shishper_aligned/april27_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"

sar_sig_files = [
    "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"
]

vv_mask_file = "bursts/2025_04_05/VV_shape.tif"
vh_mask_file = "bursts/2025_04_05/VH_shape.tif"

window_size = 10
radius = window_size // 2
min_dist = 2 * radius
band_nir = 8
lat_cutoff = 36.350

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.count_nonzero(final_mask) == 0:
        return None

    return gms[final_mask].mean()

def reproject_to_ref(src_path, ref_shape, ref_transform, ref_crs):
    with rasterio.open(src_path) as src:
        data = src.read(1)
        dst = np.zeros(ref_shape, dtype=np.float32)

        reproject(
            source=data,
            destination=dst,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=ref_transform,
            dst_crs=ref_crs,
            dst_height=ref_shape[0],
            dst_width=ref_shape[1],
            resampling=Resampling.nearest
        )
    return dst


# ---------------------------
# 3. LOAD DATA & MASKS
# ---------------------------
with rasterio.open(img_start_file) as ds:
    img_start = ds.read(band_nir).astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read(band_nir).astype(np.float32)

from rasterio.warp import reproject, Resampling

# Reproject VV and VH to PlanetScope grid
vv = reproject_to_ref(
    "bursts/2025_04_05/VV_shape.tif",
    (H, W),
    transform_ps,
    crs_ps
)

vh = reproject_to_ref(
    "bursts/2025_04_05/VH_shape.tif",
    (H, W),
    transform_ps,
    crs_ps
)


gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

sar_valid_mask = (vv > 0) | (vh > 0)
valid_mask = glacier_mask & sar_valid_mask

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. SIGNIFICANT POINTS
# ---------------------------
sig_coords = []

for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)

        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)

print(f"Significant points: {num_pts}")

# ---------------------------
# 5. RANDOM POINTS (CAN OVERLAP EACH OTHER)
# ---------------------------
rand_coords = []
attempts = 0
max_attempts = 30000

while len(rand_coords) < num_pts and attempts < max_attempts:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)

    if not valid_mask[r, c]:
        attempts += 1
        continue

    candidate = np.array([r, c])

    # ONLY enforce distance from significant points
    if np.any(np.linalg.norm(sig_coords_arr - candidate, axis=1) < min_dist):
        attempts += 1
        continue

    rand_coords.append((r, c))
    attempts += 1

print(f"Random points accepted: {len(rand_coords)}")

# ---------------------------
# 6. FSIM
# ---------------------------
sig_fsim = [
    get_masked_fsim_metrics(img_start, img_end, valid_mask, r, c, radius)
    for r, c in sig_coords
]

rand_fsim = [
    get_masked_fsim_metrics(img_start, img_end, valid_mask, r, c, radius)
    for r, c in rand_coords
]

sig_fsim = [v for v in sig_fsim if v is not None]
rand_fsim = [v for v in rand_fsim if v is not None]

# ---------------------------
# 7. STATS & PLOTS
# ---------------------------
u, p = mannwhitneyu(sig_fsim, rand_fsim, alternative="less")

print("\n" + "="*50)
print("FSIM RESULTS (Randoms may overlap)")
print("="*50)
print(f"Mean FSIM | Sig   : {np.mean(sig_fsim):.4f}")
print(f"Mean FSIM | Rand  : {np.mean(rand_fsim):.4f}")
print(f"P-value   : {p:.6e}")
print("="*50)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start, (2, 98))
ax1.imshow(img_start, cmap="gray", vmin=vmin, vmax=vmax)

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, edgecolor="red", fill=False, lw=0.8))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, edgecolor="cyan", fill=False, lw=0.8))

ax1.set_title("FSIM Windows (Randoms can overlap)")

ax2.boxplot([sig_fsim, rand_fsim], labels=["Significant", "Random"])
ax2.set_ylabel("FSIM (GMS)")
ax2.set_title("Lower = More Structural Change")

plt.tight_layout()
plt.show()
```

```python
#June 4th — Glacier-masked FSIM analysis (Glacier-only pixel windows)

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
img_start_file = "shishper_aligned/may26_shishper.tif"
img_end_file   = "shishper_aligned/june7_shishper.tif"
shapefile_path = "shishper_shapefile/shishper glacier.shp"

sar_sig_files = [
    "anomalies_tree/VV_VH_anomaly_2025-06-04.tif"
]

window_size = 50
radius = window_size // 2
min_dist = 2 * radius
band_nir = 4
lat_cutoff = 36.350

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    """Compute Gradient Magnitude Similarity (GMS) between two image patches."""
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    """Compute FSIM (GMS) score for a circular window, considering only glacier pixels."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    # Extract patches
    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    # Circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    # Glacier mask window
    glacier_win = glacier_mask[r0:r1, c0:c1]

    # Final mask: only circular pixels inside glacier
    final_mask = circ_mask & glacier_win

    # Compute FSIM using only glacier pixels (ignore everything else)
    return gms_map[final_mask].mean()

# ---------------------------
# 3. DATA LOADING & MASKING
# ---------------------------
with rasterio.open(img_start_file) as ds:
    img_start = ds.read(band_nir).astype(np.float32)
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

with rasterio.open(img_end_file) as ds:
    img_end = ds.read(band_nir).astype(np.float32)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)

# Rasterize glacier polygon
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# Latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H - 1)

# ---------------------------
# 4. MAP SAR POINTS TO PLANETSCOPE
# ---------------------------
sig_coords = []
for sar_file in sar_sig_files:
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)

        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))

sig_coords = list(set(sig_coords))
sig_coords_arr = np.array(sig_coords)
num_pts = len(sig_coords)
print(f"Verified {num_pts} significant points.")

# ---------------------------
# 5. SPATIAL REJECTION SAMPLING (allow overlap among random points)
# ---------------------------
rand_coords = []
attempts = 0
while len(rand_coords) < num_pts and attempts < 20000:
    r = random.randint(radius, row_limit - radius - 1)
    c = random.randint(radius, W - radius - 1)

    if glacier_mask[r, c]:
        dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
        # Only reject if too close to significant points
        if np.all(dist_to_sig >= min_dist):
            rand_coords.append((r, c))
    attempts += 1

# ---------------------------
# 6. CALCULATE FSIM (GMS)
# ---------------------------
sig_fsim = [
    get_masked_fsim_metrics(img_start, img_end, glacier_mask, r, c, radius)
    for r, c in sig_coords
]

rand_fsim = [
    get_masked_fsim_metrics(img_start, img_end, glacier_mask, r, c, radius)
    for r, c in rand_coords
]

# Remove any None values (in case a patch is completely outside glacier)
sig_fsim = [f for f in sig_fsim if f is not None]
rand_fsim = [f for f in rand_fsim if f is not None]

# ---------------------------
# 7. STATISTICAL TEST & VISUALIZATION
# ---------------------------
u_fsim, p_fsim = mannwhitneyu(sig_fsim, rand_fsim, alternative='less')

print("\n" + "="*45)
print("RESULTS: FEATURE SIMILARITY (FSIM / GMS)")
print("="*45)
print(f"Avg FSIM - Sig: {np.mean(sig_fsim):.4f} | Rand: {np.mean(rand_fsim):.4f}")
print(f"P-value: {p_fsim:.6f}")
print("="*45)

# Plot windows and boxplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

vmin, vmax = np.percentile(img_start, (2, 98))
ax1.imshow(img_start, cmap='gray', vmin=vmin, vmax=vmax)

for r, c in sig_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='red', fill=False, linewidth=0.8))
for r, c in rand_coords:
    ax1.add_patch(patches.Circle((c, r), radius, color='cyan', fill=False, linewidth=0.8))

ax1.set_title("Feature Analysis Windows (Glacier-Masked)")

ax2.boxplot([sig_fsim, rand_fsim], labels=['Significant', 'Random'])
ax2.set_title("FSIM (GMS) — Lower = Greater Structural Change")
ax2.set_ylabel("Similarity Score")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr


# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for FSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK & VV/VH UNION MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)


with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------

dates = []
diff_matrix = []
sig_counts = [] 


for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — FSIM difference (random - sig) mean={np.mean(iter_diffs):.4f}")
    
# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("isoforest_test.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr
from scipy.stats import wilcoxon


# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for FSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK & VV/VH UNION MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)


with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(
        sar_file, valid_mask, transform_ps, row_limit, W, radius
    )
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []

    for _ in range(n_iterations):
        rand_coords = sample_random_points(
            valid_mask, sig_coords_arr, n_sig, radius, row_limit, W
        )

        sig_avg, rand_avg = compute_avg_fsim(
            before_img, after_img, valid_mask,
            sig_coords, rand_coords, radius
        )

        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    # ---- WILCOXON TEST ----
    if len(sig_avgs) > 0:
        stat, p = wilcoxon(rand_avgs, sig_avgs, alternative='two-sided')
        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {np.mean(iter_diffs):.4f}")
        print(f"   Wilcoxon statistic = {stat:.4f}, p-value = {p:.6f}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)


# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("isoforest_test.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr
from scipy.stats import mannwhitneyu


# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for FSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK & VV/VH UNION MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)


with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(
        sar_file, valid_mask, transform_ps, row_limit, W, radius
    )
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []

    for _ in range(n_iterations):
        rand_coords = sample_random_points(
            valid_mask, sig_coords_arr, n_sig, radius, row_limit, W
        )

        sig_avg, rand_avg = compute_avg_fsim(
            before_img, after_img, valid_mask,
            sig_coords, rand_coords, radius
        )

        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

        # ---- MANN-WHITNEY U TEST ----
    if len(sig_avgs) > 0:
        stat, p = mannwhitneyu(rand_avgs, sig_avgs, alternative='greater')

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {np.mean(iter_diffs):.4f}")
        print(f"   Mann–Whitney U statistic = {stat:.4f}, p-value = {p:.2e}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)

# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("isoforest_test.png", dpi=300)
plt.show()
 
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr
from scipy.stats import mannwhitneyu


# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for FSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK & VV/VH UNION MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)


with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(
        sar_file, valid_mask, transform_ps, row_limit, W, radius
    )
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []

    for _ in range(n_iterations):
        rand_coords = sample_random_points(
            valid_mask, sig_coords_arr, n_sig, radius, row_limit, W
        )

        sig_avg, rand_avg = compute_avg_fsim(
            before_img, after_img, valid_mask,
            sig_coords, rand_coords, radius
        )

        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    # ---- COHEN'S d (PAIRED EFFECT SIZE) ----
    if len(sig_avgs) > 1:

        diffs = np.array(rand_avgs) - np.array(sig_avgs)

        mean_diff = np.mean(diffs)
        std_diff = np.std(diffs, ddof=1)

        if std_diff == 0:
            cohens_d = 0
        else:
            cohens_d = mean_diff / std_diff

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {mean_diff:.6f}")
        print(f"   Cohen's d: {cohens_d:.4f}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)


# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("isoforest_test.png", dpi=300)
plt.show()
 
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.metrics import structural_similarity as ssim

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for SSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def get_masked_ssim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    # Apply mask
    w1_masked = w1 * final_mask
    w2_masked = w2 * final_mask

    data_range = w1_masked.max() - w1_masked.min()
    if data_range == 0:
        return None

    score = ssim(w1_masked, w2_masked, data_range=data_range)
    return score


def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_ssim = [f for f in sig_ssim if f is not None]
    rand_ssim = [f for f in rand_ssim if f is not None]
    return np.mean(sig_ssim), np.mean(rand_ssim)


# ---------------------------
# 3. LOAD MASKS
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — SSIM difference (random - sig) mean={np.mean(iter_diffs):.4f}")

# ---------------------------
# 5. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("SSIM Difference (Random - Significant)")
plt.title(f"Distribution of SSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("ssim.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.metrics import structural_similarity as ssim
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for SSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def get_masked_ssim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    # Apply mask
    w1_masked = w1[final_mask]
    w2_masked = w2[final_mask]

    data_range = w1_masked.max() - w1_masked.min()
    if data_range == 0:
        return None

    score = ssim(w1_masked, w2_masked, data_range=data_range)
    return score


def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_ssim = [f for f in sig_ssim if f is not None]
    rand_ssim = [f for f in rand_ssim if f is not None]
    return np.mean(sig_ssim), np.mean(rand_ssim)


# ---------------------------
# 3. LOAD MASKS
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []


for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
    
        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    if len(sig_avgs) > 0:
        stat, p = mannwhitneyu(rand_avgs, sig_avgs, alternative='greater')

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {np.mean(iter_diffs):.4f}")
        print(f"   Mann–Whitney U statistic = {stat:.4f}, p-value = {p:.2e}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)

# ---------------------------
# 5. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("SSIM Difference (Random - Significant)")
plt.title(f"Distribution of SSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("ssim.png", dpi=300)
plt.show()
 
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.metrics import structural_similarity as ssim
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Date pairs for SSIM analysis
date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def get_masked_ssim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    # Apply mask
    w1_masked = w1[final_mask]
    w2_masked = w2[final_mask]

    data_range = w1_masked.max() - w1_masked.min()
    if data_range == 0:
        return None

    score = ssim(w1_masked, w2_masked, data_range=data_range)
    return score


def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_ssim = [get_masked_ssim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_ssim = [f for f in sig_ssim if f is not None]
    rand_ssim = [f for f in rand_ssim if f is not None]
    return np.mean(sig_ssim), np.mean(rand_ssim)


# ---------------------------
# 3. LOAD MASKS
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []


for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_ssim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
    
        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    # ---- COHEN'S d (PAIRED EFFECT SIZE) ----
    if len(sig_avgs) > 1:

        diffs = np.array(rand_avgs) - np.array(sig_avgs)

        mean_diff = np.mean(diffs)
        std_diff = np.std(diffs, ddof=1)

        if std_diff == 0:
            cohens_d = 0
        else:
            cohens_d = mean_diff / std_diff

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {mean_diff:.6f}")
        print(f"   Cohen's d: {cohens_d:.4f}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)

# ---------------------------
# 5. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("SSIM Difference (Random - Significant)")
plt.title(f"Distribution of SSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("ssim.png", dpi=300)
plt.show()
 
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. EUCLIDEAN NORMALIZED SIMILARITY
# ---------------------------

def compute_similarity(p1, p2):
    p1 = p1.flatten()
    p2 = p2.flatten()

    if len(p1) < 20:
        return None

    if np.std(p1) == 0 or np.std(p2) == 0:
        return None

    return np.corrcoef(p1, p2)[0,1]


def get_masked_similarity(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    return compute_similarity(w1[final_mask], w2[final_mask])


# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)


# ---------------------------
# 4. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 5. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — Similarity difference (random - sig) mean={np.mean(iter_diffs):.4f}")


# ---------------------------
# 6. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("Similarity Difference (Random - Significant)")
plt.title(f"Distribution of Similarity Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("z.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. EUCLIDEAN NORMALIZED SIMILARITY
# ---------------------------

def compute_similarity(p1, p2):
    p1 = p1.flatten()
    p2 = p2.flatten()

    if len(p1) < 20:
        return None

    if np.std(p1) == 0 or np.std(p2) == 0:
        return None

    return np.corrcoef(p1, p2)[0,1]


def get_masked_similarity(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    return compute_similarity(w1[final_mask], w2[final_mask])


# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)


# ---------------------------
# 4. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 5. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    sig_avgs = []
    rand_avgs = []
    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
    
        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    if len(sig_avgs) > 0:
        stat, p = mannwhitneyu(rand_avgs, sig_avgs, alternative='greater')

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {np.mean(iter_diffs):.4f}")
        print(f"   Mann–Whitney U statistic = {stat:.4f}, p-value = {p:.2e}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)


# ---------------------------
# 6. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("Similarity Difference (Random - Significant)")
plt.title(f"Distribution of ZNCC Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("z.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from scipy.ndimage import sobel
from numpy.linalg import norm

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

n_bins = 8   # orientation bins for LESH

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. LESH FUNCTION
# ---------------------------

def compute_lesh_descriptor(patch, mask):

    # Compute gradients
    gx = sobel(patch, axis=0)
    gy = sobel(patch, axis=1)

    magnitude = np.sqrt(gx**2 + gy**2)
    orientation = np.arctan2(gy, gx)

    # Only use valid pixels
    magnitude = magnitude[mask]
    orientation = orientation[mask]

    if len(magnitude) < 20:
        return None

    # Convert orientation to [0, pi]
    orientation = np.mod(orientation, np.pi)

    # Histogram weighted by magnitude
    hist, _ = np.histogram(
        orientation,
        bins=n_bins,
        range=(0, np.pi),
        weights=magnitude
    )

    # Normalize histogram
    if np.sum(hist) == 0:
        return None

    hist = hist / np.sum(hist)

    return hist


def get_masked_lesh(img1, img2, valid_mask, r, c, rad):

    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    d1 = compute_lesh_descriptor(w1, final_mask)
    d2 = compute_lesh_descriptor(w2, final_mask)

    if d1 is None or d2 is None:
        return None

    # Cosine similarity (0–1)
    sim = np.dot(d1, d2) / (norm(d1) * norm(d2))
    return sim


def compute_avg_lesh(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_lesh(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_lesh(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)



# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)


# ---------------------------
# 4. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)



gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_lesh(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — LESH difference (random - sig) mean={np.mean(iter_diffs):.4f}")


# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("LESH Difference (Random - Significant)")
plt.title(f"Distribution of LESH Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("lesh.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from sklearn.metrics import mutual_info_score
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Number of bins for MI histogram
n_bins = 32

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. MUTUAL INFORMATION FUNCTION
# ---------------------------

def get_masked_mi(img1, img2, valid_mask, r, c, rad):

    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    vals1 = w1[final_mask]
    vals2 = w2[final_mask]

    # Normalize to bins
    vals1 = np.digitize(vals1, np.histogram(vals1, bins=n_bins)[1])
    vals2 = np.digitize(vals2, np.histogram(vals2, bins=n_bins)[1])

    mi = mutual_info_score(vals1, vals2)

    return mi


def compute_avg_mi(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_mi(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_mi(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)



# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)


# ---------------------------
# 4. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)



gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"


# ---------------------------
# 3–4. EVERYTHING ELSE UNCHANGED
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 5. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:

    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)

    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_mi(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
    
        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    if len(sig_avgs) > 0:
        stat, p = mannwhitneyu(rand_avgs, sig_avgs, alternative='greater')

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {np.mean(iter_diffs):.4f}")
        print(f"   Mann–Whitney U statistic = {stat:.4f}, p-value = {p:.2e}\n")

    dates.append(label)
    diff_matrix.append(iter_diffs)


# ---------------------------
# 6. BOXPLOT
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--')
plt.xticks(rotation=45)
plt.ylabel("Mutual Information Difference (Random - Significant)")
plt.title(f"Distribution of MI Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("mutual_information.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from sklearn.metrics import mutual_info_score
from scipy.stats import mannwhitneyu

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# Number of bins for MI histogram
n_bins = 32

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. MUTUAL INFORMATION FUNCTION
# ---------------------------

def get_masked_mi(img1, img2, valid_mask, r, c, rad):

    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    vals1 = w1[final_mask]
    vals2 = w2[final_mask]

    # Normalize to bins
    vals1 = np.digitize(vals1, np.histogram(vals1, bins=n_bins)[1])
    vals2 = np.digitize(vals2, np.histogram(vals2, bins=n_bins)[1])

    mi = mutual_info_score(vals1, vals2)

    return mi


def compute_avg_mi(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_mi(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_mi(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)



# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)


# ---------------------------
# 4. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)



gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"


# ---------------------------
# 3–4. EVERYTHING ELSE UNCHANGED
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 5. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:

    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)

    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_mi(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
    
        if not np.isnan(sig_avg) and not np.isnan(rand_avg):
            sig_avgs.append(sig_avg)
            rand_avgs.append(rand_avg)
            iter_diffs.append(rand_avg - sig_avg)

    # ---- COHEN'S d (PAIRED EFFECT SIZE) ----
    if len(sig_avgs) > 1:

        diffs = np.array(rand_avgs) - np.array(sig_avgs)

        mean_diff = np.mean(diffs)
        std_diff = np.std(diffs, ddof=1)

        if std_diff == 0:
            cohens_d = 0
        else:
            cohens_d = mean_diff / std_diff

        print(f"{label} — n={n_sig}")
        print(f"   Mean Difference (Random - Sig): {mean_diff:.6f}")
        print(f"   Cohen's d: {cohens_d:.4f}\n")


    dates.append(label)
    diff_matrix.append(iter_diffs)


# ---------------------------
# 6. BOXPLOT
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--')
plt.xticks(rotation=45)
plt.ylabel("Mutual Information Difference (Random - Significant)")
plt.title(f"Distribution of MI Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("mutual_information.png", dpi=300)
plt.show()
```

```python

from rasterio.warp import reproject, Resampling
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100


# ---------------------------
# 2. HOG MODEL FUNCTION
# ---------------------------

def compute_hog_descriptor(patch, n_bins=9):

    dx = scharr(patch, axis=0)
    dy = scharr(patch, axis=1)

    magnitude = np.sqrt(dx**2 + dy**2)
    orientation = np.arctan2(dy, dx)  # radians
    orientation = np.degrees(orientation) % 180  # unsigned gradients

    magnitude = magnitude.flatten()
    orientation = orientation.flatten()

    valid = ~np.isnan(magnitude)
    magnitude = magnitude[valid]
    orientation = orientation[valid]

    if len(magnitude) < 20:
        return None

    hist, _ = np.histogram(
        orientation,
        bins=n_bins,
        range=(0, 180),
        weights=magnitude
    )

    hist = hist.astype(np.float32)
    hist /= (np.linalg.norm(hist) + 1e-10)

    return hist


def get_masked_hog(img1, img2, valid_mask, r, c, rad):

    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    w1_masked = np.where(final_mask, w1, np.nan)
    w2_masked = np.where(final_mask, w2, np.nan)

    h1 = compute_hog_descriptor(w1_masked)
    h2 = compute_hog_descriptor(w2_masked)

    if h1 is None or h2 is None:
        return None

    return np.linalg.norm(h1 - h2)  # L2 distance


def compute_avg_hog(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_hog(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_hog(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)




# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

   

    return loss_h + loss_v


# ---------------------------
# 3. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:

    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)

    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_hog(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)

    print(f"{label} — {n_sig} sig points — Histogram of Gradient difference (random - sig) mean={np.mean(iter_diffs):.4f}")


# ---------------------------
# 5. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("HoG Difference (Random - Significant)")
plt.title(f"Distribution of HoG Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("hog.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 25
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_forest/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif","anomalies_forest/VV_VH_anomaly_2025-10-26.tif"),
]

n_iterations = 100



# ---------------------------
# 2. MONOTONIC CORRELATION FUNCTION (PCE)
# ---------------------------

def get_masked_pce(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win

    if np.sum(final_mask) < 20:
        return None

    vals1 = w1[final_mask]
    vals2 = w2[final_mask]

    # Normalize patches individually (Zero-mean, Unit-variance)
    # δ_t/σ_t and δ_i,j/σ_i,j as per paper
    std1, std2 = np.std(vals1), np.std(vals2)
    if std1 == 0 or std2 == 0:
        return None
        
    norm1 = (vals1 - np.mean(vals1)) / std1
    norm2 = (vals2 - np.mean(vals2)) / std2

    # Calculate Normalized Euclidean Distance (Delta)
    # Delta = sum((norm1 - norm2)^2)
    delta = np.sum((norm1 - norm2)**2)

    # Convert to Correlation Coefficient (rho)
    # rho = 1 - 0.5 * (delta / N) where N is normalized per-pixel
    # Paper uses direct summation in Eq 5: rho = 1 - 1/2 * sum(diff^2)
    # Because we use vectors of different lengths (masked), we divide by N to normalize magnitude
    rho = 1 - 0.5 * (delta / len(vals1))

    return abs(rho) # Using absolute value as per previous discussion


def compute_avg_pce(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_pce(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_pce(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    return np.mean(sig_vals), np.mean(rand_vals)




# ---------------------------
# 3. HELPER FUNCTIONS (UNCHANGED)
# ---------------------------

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))


def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_similarity(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_vals = [get_masked_similarity(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

   

    return loss_h + loss_v


# ---------------------------
# 3. LOAD MASKS (UNCHANGED)
# ---------------------------

with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H)

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data = reproject_to_target(ds_vv.read(1), ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data = reproject_to_target(ds_vh.read(1), ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------

dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:

    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)

    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_pce(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)

    print(f"{label} — {n_sig} sig points — Histogram of PCE difference (random - sig) mean={np.mean(iter_diffs):.4f}")


# ---------------------------
# 5. BOXPLOT
# ---------------------------

plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("PCE Difference (Random - Significant)")
plt.title(f"Distribution of PCE Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("pce.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 30
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

# List of date tuples: (label, before_image, after_image, SAR significant file)
date_pairs = [
    #("Apr 5", "shishper_aligned/april2_shishper.tif", "shishper_aligned/april17_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "wilks/wilks_points_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "wilks/wilks_points_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "wilks/wilks_points_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "wilks/wilks_points_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/june7_shishper.tif", "wilks/wilks_points_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june24_shishper.tif", "wilks/wilks_points_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/july5_shishper.tif", "wilks/wilks_points_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july13_shishper.tif", "wilks/wilks_points_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif", "wilks/wilks_points_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif", "wilks/wilks_points_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif", "wilks/wilks_points_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif", "wilks/wilks_points_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif", "wilks/wilks_points_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif", "wilks/wilks_points_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif", "wilks/wilks_points_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/sep21_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif", "wilks/wilks_points_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif", "wilks/wilks_points_2025-10-26.tif"),
    
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    valid_win = valid_mask[r0:r1, c0:c1]
    final_mask = circ_mask & valid_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if valid_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(valid_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if valid_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, valid_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK & VV/VH UNION MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# --- Single SAR VV/VH file for valid data mask ---
vv_file_single = "bursts/2025_04_05/VV_shape.tif"
vh_file_single = "bursts/2025_04_05/VH_shape.tif"

# --- Reproject VV/VH to match Planetscope raster ---
def reproject_to_target(src_array, src_transform, src_crs, target_shape, target_transform, target_crs):
    dst_array = np.zeros(target_shape, dtype=src_array.dtype)
    reproject(
        source=src_array,
        destination=dst_array,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=target_transform,
        dst_crs=target_crs,
        resampling=Resampling.nearest
    )
    return dst_array

with rasterio.open(vv_file_single) as ds_vv:
    vv_data_orig = ds_vv.read(1)
    vv_data = reproject_to_target(vv_data_orig, ds_vv.transform, ds_vv.crs, (H, W), transform_ps, crs_ps)

with rasterio.open(vh_file_single) as ds_vh:
    vh_data_orig = ds_vh.read(1)
    vh_data = reproject_to_target(vh_data_orig, ds_vh.transform, ds_vh.crs, (H, W), transform_ps, crs_ps)

# Now shapes match, compute union mask
valid_mask = glacier_mask & ((vv_data > 0) | (vh_data > 0))


# ---------------------------
# 4. LOOP OVER DATES + RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []
sig_counts = []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, valid_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []
    for _ in range(n_iterations):
        rand_coords = sample_random_points(valid_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, valid_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — FSIM difference (random - sig) mean={np.mean(iter_diffs):.4f}")

# ---------------------------
# 5. BOXPLOT WITH POINT COUNTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("wilks.png", dpi=300)
plt.show()
```

```python
import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 10
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"

date_pairs = [
    #("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_tree/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "change/VV_VH_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "change/VV_VH_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "change/VV_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "change/VV_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif", "change/VV_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif", "change/VV_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif", "change/VV_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif", "change/VV_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif", "change/VV_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif", "change/VV_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif", "change/VV_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif", "change/VV_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif", "change/VV_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif", "change/VV_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif", "change/VV_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif", "change/VV_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif", "change/VV_2025-10-26.tif"),
]

n_iterations = 100  # Number of random draws

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    dx = scharr(w1, axis=0)
    dy = scharr(w1, axis=1)
    g1 = np.sqrt(dx**2 + dy**2)
    dx = scharr(w2, axis=0)
    dy = scharr(w2, axis=1)
    g2 = np.sqrt(dx**2 + dy**2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2
    glacier_win = glacier_mask[r0:r1, c0:c1]
    final_mask = circ_mask & glacier_win
    if np.sum(final_mask) < 20:
        return None
    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius):
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)
        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(glacier_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    """
    Random points may overlap each other but cannot overlap significant points.
    """
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if glacier_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                # allow overlap among random points
                rand_coords.append((r, c))
        attempts += 1
    return rand_coords


def compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. LOOP OVER DATES + 100 RANDOM ITERATIONS
# ---------------------------
dates = []
diff_matrix = []  # list of lists: each inner list = 100 iterations difference for one date
sig_counts = []   # store number of significant points per date

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    sig_coords = map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)
    n_sig = len(sig_coords)
    sig_counts.append(n_sig)

    iter_diffs = []

    for _ in range(n_iterations):
        rand_coords = sample_random_points(glacier_mask, sig_coords_arr, n_sig, radius, row_limit, W)
        sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius)
        iter_diffs.append(rand_avg - sig_avg)  # difference: random - significant

    dates.append(label)
    diff_matrix.append(iter_diffs)
    print(f"{label} — {n_sig} sig points — FSIM difference (random - sig) over {n_iterations} runs: mean={np.mean(iter_diffs):.4f}")

# ---------------------------
# 5. BOXPLOT WITH NUMBER OF POINTS
# ---------------------------
plt.figure(figsize=(14,6))
labels_with_counts = [f"{d}\n(n={c})" for d,c in zip(dates, sig_counts)]
plt.boxplot(diff_matrix, labels=labels_with_counts, showfliers=False)
plt.axhline(0, color='red', linestyle='--', label='No Difference')
plt.xticks(rotation=45)
plt.ylabel("FSIM Difference (Random - Significant)")
plt.title(f"Distribution of FSIM Difference over {n_iterations} Random Runs")
plt.grid(True)
plt.tight_layout()
plt.savefig("fsim_esa_change.png", dpi=300)
plt.show()
```

```python
# ---------------------------
# Glacier-Masked FSIM Time Series Analysis
# ---------------------------

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 70
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

# Shapefile of glacier
shapefile_path = "shishper_shapefile/shishper glacier.shp"

# List of date tuples: (label, before_image, after_image, SAR significant file)
date_pairs = [
    ("Apr 5", "shishper_aligned/april2_shishper.tif", "shishper_aligned/april17_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned/april17_shishper.tif", "shishper_aligned/april27_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned/april27_shishper.tif", "shishper_aligned/may6_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-04-29.tif"),
    ("May 11", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-05-11.tif"),
    ("June 4", "shishper_aligned/may26_shishper.tif", "shishper_aligned/june7_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-06-04.tif"),
    ("June 16", "shishper_aligned/june7_shishper.tif", "shishper_aligned/june24_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-06-16.tif"),
    ("June 28", "shishper_aligned/june24_shishper.tif", "shishper_aligned/july5_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-06-28.tif"),
    ("July 10", "shishper_aligned/july5_shishper.tif", "shishper_aligned/july13_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-07-10.tif"),
    ("July 22", "shishper_aligned/july13_shishper.tif", "shishper_aligned/july21_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned/july21_shishper.tif", "shishper_aligned/aug12_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned/aug12_shishper.tif", "shishper_aligned/aug19_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned/aug19_shishper.tif", "shishper_aligned/aug26_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned/aug26_shishper.tif", "shishper_aligned/sep10_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned/sep10_shishper.tif", "shishper_aligned/sep21_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned/sep28_shishper.tif", "shishper_aligned/oct12_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned/oct12_shishper.tif", "shishper_aligned/oct15_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned/oct15_shishper.tif", "shishper_aligned/oct31_shishper.tif", "max_ratios_significant/ratio_significant_99_2025-10-26.tif"),
    
]

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def calculate_gms(w1, w2):
    """Calculate Gradient Magnitude Similarity (GMS) map between two windows."""
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    """Calculate FSIM (GMS) within a circular window masked by glacier."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    # Circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    # Glacier mask window
    glacier_win = glacier_mask[r0:r1, c0:c1]

    # Final mask = circle ∩ glacier
    final_mask = circ_mask & glacier_win

    if np.sum(final_mask) < 20:  # discard tiny glacier areas
        return None

    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius):
    """Map SAR significant points to the high-res satellite image."""
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(glacier_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    """Spatial rejection sampling to get random control points inside glacier."""
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if glacier_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                if not rand_coords or np.all(np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist):
                    rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK
# ---------------------------
# Load one image to get size & transform
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# Latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------
dates = []
sig_avg_list = []
rand_avg_list = []

for label, before_file, after_file, sar_file in date_pairs:
    # Load before/after images
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    # Map SAR points
    sig_coords = map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)

    # Sample random points
    rand_coords = sample_random_points(glacier_mask, sig_coords_arr, len(sig_coords), radius, row_limit, W)

    # Compute averages
    sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius)

    dates.append(label)
    sig_avg_list.append(sig_avg)
    rand_avg_list.append(rand_avg)

    print(f"{label} — Avg FSIM: Sig={sig_avg:.4f}, Rand={rand_avg:.4f}")

# ---------------------------
# 5. PLOT LINE CHART
# ---------------------------
plt.figure(figsize=(12,6))
plt.plot(dates, sig_avg_list, marker='o', label='Significant Points')
plt.plot(dates, rand_avg_list, marker='o', label='Random Points')
plt.xlabel("Date of SAR Acquisition")
plt.ylabel("Average FSIM (Higher = Less Change)")
plt.title("Glacier Feature Similarity Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ---------------------------
# Glacier-Masked FSIM Time Series Analysis
# ---------------------------

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 70
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

# Shapefile of glacier
shapefile_path = "shishper_shapefile/shishper glacier.shp"

# List of date tuples: (label, before_image, after_image, SAR significant file)
date_pairs = [
    ("Apr 5", "shishper_aligned/april2_shishper.tif", "shishper_aligned/april17_shishper.tif", "anomalies/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned/april17_shishper.tif", "shishper_aligned/april27_shishper.tif", "anomalies/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned/april27_shishper.tif", "shishper_aligned/may6_shishper.tif", "anomalies/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "anomalies/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "anomalies/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned/may26_shishper.tif", "shishper_aligned/june7_shishper.tif", "anomalies/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned/june7_shishper.tif", "shishper_aligned/june24_shishper.tif", "anomalies/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned/june24_shishper.tif", "shishper_aligned/july5_shishper.tif", "anomalies/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned/july5_shishper.tif", "shishper_aligned/july13_shishper.tif", "anomalies/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned/july13_shishper.tif", "shishper_aligned/july21_shishper.tif", "anomalies/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned/july21_shishper.tif", "shishper_aligned/aug12_shishper.tif", "anomalies/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned/aug12_shishper.tif", "shishper_aligned/aug19_shishper.tif", "anomalies/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned/aug19_shishper.tif", "shishper_aligned/aug26_shishper.tif", "anomalies/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned/aug26_shishper.tif", "shishper_aligned/sep10_shishper.tif", "anomalies/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned/sep10_shishper.tif", "shishper_aligned/sep21_shishper.tif", "anomalies/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned/sep28_shishper.tif", "shishper_aligned/oct12_shishper.tif", "anomalies/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned/oct12_shishper.tif", "shishper_aligned/oct15_shishper.tif", "anomalies/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned/oct15_shishper.tif", "shishper_aligned/oct31_shishper.tif", "anomalies/VV_VH_anomaly_2025-10-26.tif"),
    
]

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def calculate_gms(w1, w2):
    """Calculate Gradient Magnitude Similarity (GMS) map between two windows."""
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    """Calculate FSIM (GMS) within a circular window masked by glacier."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    # Circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    # Glacier mask window
    glacier_win = glacier_mask[r0:r1, c0:c1]

    # Final mask = circle ∩ glacier
    final_mask = circ_mask & glacier_win

    if np.sum(final_mask) < 20:  # discard tiny glacier areas
        return None

    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius):
    """Map SAR significant points to the high-res satellite image."""
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(glacier_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    """Spatial rejection sampling to get random control points inside glacier."""
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if glacier_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                if not rand_coords or np.all(np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist):
                    rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK
# ---------------------------
# Load one image to get size & transform
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# Latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------
dates = []
sig_avg_list = []
rand_avg_list = []

for label, before_file, after_file, sar_file in date_pairs:
    # Load before/after images
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    # Map SAR points
    sig_coords = map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)

    # Sample random points
    rand_coords = sample_random_points(glacier_mask, sig_coords_arr, len(sig_coords), radius, row_limit, W)

    # Compute averages
    sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius)

    dates.append(label)
    sig_avg_list.append(sig_avg)
    rand_avg_list.append(rand_avg)

    print(f"{label} — Avg FSIM: Sig={sig_avg:.4f}, Rand={rand_avg:.4f}")

# ---------------------------
# 5. PLOT LINE CHART
# ---------------------------
plt.figure(figsize=(12,6))
plt.plot(dates, sig_avg_list, marker='o', label='Significant Points')
plt.plot(dates, rand_avg_list, marker='o', label='Random Points')
plt.xlabel("Date of SAR Acquisition")
plt.ylabel("Average FSIM (Higher = Less Change)")
plt.title("Glacier Feature Similarity Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ---------------------------
# Glacier + SAR-valid Masked FSIM Time Series Analysis
# ---------------------------

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
from rasterio import features
from rasterio.warp import reproject, Resampling
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 70
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

shapefile_path = "shishper_shapefile/shishper glacier.shp"
vv_mask_file = "bursts/2025_04_05/VV_shape.tif"   # SAR-valid mask

date_pairs = [
    ("Apr 5", "shishper_aligned_nn/april2_shishper.tif", "shishper_aligned_nn/april17_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned_nn/april17_shishper.tif", "shishper_aligned_nn/april27_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned_nn/april27_shishper.tif", "shishper_aligned_nn/may6_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned_nn/may6_shishper.tif", "shishper_aligned_nn/may26_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned_nn/may26_shishper.tif", "shishper_aligned_nn/june7_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned_nn/june7_shishper.tif", "shishper_aligned_nn/june24_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned_nn/june24_shishper.tif", "shishper_aligned_nn/july5_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned_nn/july5_shishper.tif", "shishper_aligned_nn/july13_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned_nn/july13_shishper.tif", "shishper_aligned_nn/july21_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned_nn/july21_shishper.tif", "shishper_aligned_nn/aug12_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned_nn/aug12_shishper.tif", "shishper_aligned_nn/aug19_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned_nn/aug19_shishper.tif", "shishper_aligned_nn/aug26_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned_nn/aug26_shishper.tif", "shishper_aligned_nn/sep10_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned_nn/sep10_shishper.tif", "shishper_aligned_nn/sep21_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned_nn/sep28_shishper.tif", "shishper_aligned_nn/oct12_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned_nn/oct12_shishper.tif", "shishper_aligned_nn/oct15_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned_nn/oct15_shishper.tif", "shishper_aligned_nn/oct31_shishper.tif", "anomalies_rolling/VV_VH_anomaly_2025-10-26.tif"),
    
]

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------
def calculate_gms(w1, w2):
    def grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = grad(w1)
    g2 = grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)


def get_masked_fsim(img1, img2, glacier_mask, vv_valid_mask, r, c, rad):
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1

    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms = calculate_gms(w1, w2)

    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ = x**2 + y**2 <= rad**2

    mask = (
        circ &
        glacier_mask[r0:r1, c0:c1] &
        vv_valid_mask[r0:r1, c0:c1]
    )

    if np.count_nonzero(mask) < 20:
        return None

    return gms[mask].mean()


def map_sar_points(sar_file, glacier_mask, vv_valid_mask, transform_ps, H, W, radius):
    pts = []
    with rasterio.open(sar_file) as ds:
        m = ds.read(1) > 0
        r_s, c_s = np.where(m)
        xs, ys = rasterio.transform.xy(ds.transform, r_s, c_s, offset="center")
        cols, rows = ~transform_ps * (xs, ys)
        rows = np.round(rows).astype(int)
        cols = np.round(cols).astype(int)

        for r, c in zip(rows, cols):
            if radius <= r < H-radius and radius <= c < W-radius:
                if glacier_mask[r, c] and vv_valid_mask[r, c]:
                    pts.append((r, c))
    return list(set(pts))


def sample_random(glacier_mask, vv_valid_mask, sig_arr, n, H, W, radius):
    out = []
    tries = 0
    while len(out) < n and tries < 30000:
        r = random.randint(radius, H-radius-1)
        c = random.randint(radius, W-radius-1)
        if glacier_mask[r, c] and vv_valid_mask[r, c]:
            # Only check that this exact point hasn't already been selected
            if (r, c) not in out:
                out.append((r, c))
        tries += 1
    return out

# ---------------------------
# 3. LOAD GRID + MASKS
# ---------------------------
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

# Glacier mask
gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    default_value=1,
    all_touched=True
).astype(bool)

# SAR-valid mask reprojection
vv_valid_mask = np.zeros((H, W), dtype=np.uint8)
with rasterio.open(vv_mask_file) as ds_vv:
    reproject(
        source=ds_vv.read(1),
        destination=vv_valid_mask,
        src_transform=ds_vv.transform,
        src_crs=ds_vv.crs,
        dst_transform=transform_ps,
        dst_crs=crs_ps,
        resampling=Resampling.nearest
    )
vv_valid_mask = vv_valid_mask > 0

analysis_mask = glacier_mask & vv_valid_mask

# ---------------------------
# 4. TIME SERIES LOOP
# ---------------------------
dates, sig_avg, rand_avg = [], [], []

for label, before_file, after_file, sar_file in date_pairs:
    with rasterio.open(before_file) as ds:
        before = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after = ds.read(band_nir).astype(np.float32)

    sig_pts = map_sar_points(
        sar_file, glacier_mask, vv_valid_mask,
        transform_ps, H, W, radius
    )
    sig_arr = np.array(sig_pts)

    rand_pts = sample_random(
        glacier_mask, vv_valid_mask,
        sig_arr, len(sig_pts),
        H, W, radius
    )

    sig_vals = [
        get_masked_fsim(before, after, glacier_mask, vv_valid_mask, r, c, radius)
        for r, c in sig_pts
    ]
    rand_vals = [
        get_masked_fsim(before, after, glacier_mask, vv_valid_mask, r, c, radius)
        for r, c in rand_pts
    ]

    sig_vals = [v for v in sig_vals if v is not None]
    rand_vals = [v for v in rand_vals if v is not None]

    dates.append(label)
    sig_avg.append(np.mean(sig_vals))
    rand_avg.append(np.mean(rand_vals))

    print(f"{label}: FSIM sig={sig_avg[-1]:.4f}, rand={rand_avg[-1]:.4f}")

# ---------------------------
# 5. PLOT
# ---------------------------
plt.figure(figsize=(12,6))
plt.plot(dates, sig_avg, marker='o', label="SAR Anomaly")
plt.plot(dates, rand_avg, marker='o', label="Random Control")
plt.ylabel("Average FSIM (lower = more change)")
plt.xlabel("SAR Acquisition Date")
plt.title("Glacier FSIM Time Series (Glacier ∩ SAR-valid)")
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ---------------------------
# Glacier-Masked FSIM Time Series Analysis
# ---------------------------

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 100
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

# Shapefile of glacier
shapefile_path = "shishper_shapefile/shishper glacier.shp"

# List of date tuples: (label, before_image, after_image, SAR significant file)
date_pairs = [
    ("Apr 5", "shishper_aligned/april2_shishper.tif", "shishper_aligned/april17_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned/april17_shishper.tif", "shishper_aligned/april27_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned/april27_shishper.tif", "shishper_aligned/may6_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-04-29.tif"),
    ("May 11", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-05-11.tif"),
    ("May 23", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-05-23.tif"),
    ("June 4", "shishper_aligned/may26_shishper.tif", "shishper_aligned/june7_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-06-04.tif"),
    ("June 16", "shishper_aligned/june7_shishper.tif", "shishper_aligned/june24_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-06-16.tif"),
    ("June 28", "shishper_aligned/june24_shishper.tif", "shishper_aligned/july5_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-06-28.tif"),
    ("July 10", "shishper_aligned/july5_shishper.tif", "shishper_aligned/july13_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-07-10.tif"),
    ("July 22", "shishper_aligned/july13_shishper.tif", "shishper_aligned/july21_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned/july21_shishper.tif", "shishper_aligned/aug12_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned/aug12_shishper.tif", "shishper_aligned/aug19_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned/aug19_shishper.tif", "shishper_aligned/aug26_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned/aug26_shishper.tif", "shishper_aligned/sep10_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned/sep10_shishper.tif", "shishper_aligned/sep21_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned/sep28_shishper.tif", "shishper_aligned/oct12_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned/oct12_shishper.tif", "shishper_aligned/oct15_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned/oct15_shishper.tif", "shishper_aligned/oct31_shishper.tif", "anomalies_VV_raw/VV_anomaly_2025-10-26.tif"),
    
]

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def calculate_gms(w1, w2):
    """Calculate Gradient Magnitude Similarity (GMS) map between two windows."""
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    """Calculate FSIM (GMS) within a circular window masked by glacier."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    # Circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    # Glacier mask window
    glacier_win = glacier_mask[r0:r1, c0:c1]

    # Final mask = circle ∩ glacier
    final_mask = circ_mask & glacier_win

    if np.sum(final_mask) < 20:  # discard tiny glacier areas
        return None

    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius):
    """Map SAR significant points to the high-res satellite image."""
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(glacier_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    """Spatial rejection sampling to get random control points inside glacier."""
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if glacier_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                if not rand_coords or np.all(np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist):
                    rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK
# ---------------------------
# Load one image to get size & transform
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# Latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------
dates = []
sig_avg_list = []
rand_avg_list = []

for label, before_file, after_file, sar_file in date_pairs:
    # Load before/after images
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    # Map SAR points
    sig_coords = map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)

    # Sample random points
    rand_coords = sample_random_points(glacier_mask, sig_coords_arr, len(sig_coords), radius, row_limit, W)

    # Compute averages
    sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius)

    dates.append(label)
    sig_avg_list.append(sig_avg)
    rand_avg_list.append(rand_avg)

    print(f"{label} — Avg FSIM: Sig={sig_avg:.4f}, Rand={rand_avg:.4f}")

# ---------------------------
# 5. PLOT LINE CHART
# ---------------------------
plt.figure(figsize=(12,6))
plt.plot(dates, sig_avg_list, marker='o', label='Significant Points')
plt.plot(dates, rand_avg_list, marker='o', label='Random Points')
plt.xlabel("Date of SAR Acquisition")
plt.ylabel("Average FSIM (Higher = Less Change)")
plt.title("Glacier Feature Similarity Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ---------------------------
# Glacier-Masked FSIM Time Series Analysis
# ---------------------------

import numpy as np
import rasterio
import geopandas as gpd
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from rasterio import features
from skimage.filters import scharr

# ---------------------------
# 1. PARAMETERS & PATHS
# ---------------------------
band_nir = 8
window_size = 20
radius = window_size // 2
min_dist = 2 * radius
lat_cutoff = 36.350

# Shapefile of glacier
shapefile_path = "shishper_shapefile/shishper glacier.shp"

# List of date tuples: (label, before_image, after_image, SAR significant file)
date_pairs = [
    ("Apr 5", "shishper_aligned/april2_shishper.tif", "shishper_aligned/april17_shishper.tif", "thresh/above_thresh_2025-04-05.tif"),
    ("Apr 17", "shishper_aligned/april17_shishper.tif", "shishper_aligned/april27_shishper.tif", "thresh/above_thresh_2025-04-17.tif"),
    ("Apr 29", "shishper_aligned/april27_shishper.tif", "shishper_aligned/may6_shishper.tif", "thresh/above_thresh_2025-04-29.tif"),
    ("May 11", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "thresh/above_thresh_2025-05-11.tif"),
    ("May 23", "shishper_aligned/may6_shishper.tif", "shishper_aligned/may26_shishper.tif", "thresh/above_thresh_2025-05-23.tif"),
    ("June 4", "shishper_aligned/may26_shishper.tif", "shishper_aligned/june7_shishper.tif", "thresh/above_thresh_2025-06-04.tif"),
    ("June 16", "shishper_aligned/june7_shishper.tif", "shishper_aligned/june24_shishper.tif", "thresh/above_thresh_2025-06-16.tif"),
    ("June 28", "shishper_aligned/june24_shishper.tif", "shishper_aligned/july5_shishper.tif", "thresh/above_thresh_2025-06-28.tif"),
    ("July 10", "shishper_aligned/july5_shishper.tif", "shishper_aligned/july13_shishper.tif", "thresh/above_thresh_2025-07-10.tif"),
    ("July 22", "shishper_aligned/july13_shishper.tif", "shishper_aligned/july21_shishper.tif", "thresh/above_thresh_2025-07-22.tif"),
    ("Aug 3", "shishper_aligned/july21_shishper.tif", "shishper_aligned/aug12_shishper.tif", "thresh/above_thresh_2025-08-03.tif"),
    ("Aug 15", "shishper_aligned/aug12_shishper.tif", "shishper_aligned/aug19_shishper.tif", "thresh/above_thresh_2025-08-15.tif"),
    ("Aug 27", "shishper_aligned/aug19_shishper.tif", "shishper_aligned/aug26_shishper.tif", "thresh/above_thresh_2025-08-27.tif"),
    ("Sep 8", "shishper_aligned/aug26_shishper.tif", "shishper_aligned/sep10_shishper.tif", "thresh/above_thresh_2025-09-08.tif"),
    ("Sep 20", "shishper_aligned/sep10_shishper.tif", "shishper_aligned/sep21_shishper.tif", "thresh/above_thresh_2025-09-20.tif"),
    ("Oct 2", "shishper_aligned/sep28_shishper.tif", "shishper_aligned/oct12_shishper.tif", "thresh/above_thresh_2025-10-02.tif"),
    ("Oct 14", "shishper_aligned/oct12_shishper.tif", "shishper_aligned/oct15_shishper.tif", "thresh/above_thresh_2025-10-14.tif"),
    ("Oct 26", "shishper_aligned/oct15_shishper.tif", "shishper_aligned/oct31_shishper.tif", "thresh/above_thresh_2025-10-26.tif"),
    
]

# ---------------------------
# 2. HELPER FUNCTIONS
# ---------------------------

def calculate_gms(w1, w2):
    """Calculate Gradient Magnitude Similarity (GMS) map between two windows."""
    def get_grad(img):
        dx = scharr(img, axis=0)
        dy = scharr(img, axis=1)
        return np.sqrt(dx**2 + dy**2)

    g1 = get_grad(w1)
    g2 = get_grad(w2)
    c = 0.002
    return (2 * g1 * g2 + c) / (g1**2 + g2**2 + c)

def get_masked_fsim_metrics(img1, img2, glacier_mask, r, c, rad):
    """Calculate FSIM (GMS) within a circular window masked by glacier."""
    r0, r1 = r - rad, r + rad + 1
    c0, c1 = c - rad, c + rad + 1
    if r0 < 0 or r1 > img1.shape[0] or c0 < 0 or c1 > img1.shape[1]:
        return None

    w1 = img1[r0:r1, c0:c1]
    w2 = img2[r0:r1, c0:c1]
    gms_map = calculate_gms(w1, w2)

    # Circular mask
    y, x = np.ogrid[-rad:rad+1, -rad:rad+1]
    circ_mask = x**2 + y**2 <= rad**2

    # Glacier mask window
    glacier_win = glacier_mask[r0:r1, c0:c1]

    # Final mask = circle ∩ glacier
    final_mask = circ_mask & glacier_win

    if np.sum(final_mask) < 20:  # discard tiny glacier areas
        return None

    return gms_map[final_mask].mean()

def map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius):
    """Map SAR significant points to the high-res satellite image."""
    sig_coords = []
    with rasterio.open(sar_file) as ds_sar:
        mask = ds_sar.read(1) > 0
        r_sar, c_sar = np.where(mask)
        xs, ys = rasterio.transform.xy(ds_sar.transform, r_sar, c_sar, offset='center')
        cols_p, rows_p = ~transform_ps * (xs, ys)
        rows_p = np.round(rows_p).astype(int)
        cols_p = np.round(cols_p).astype(int)

        for r, c in zip(rows_p, cols_p):
            if (radius <= r < row_limit - radius) and (radius <= c < W - radius):
                if glacier_mask[r, c]:
                    sig_coords.append((r, c))
    return list(set(sig_coords))

def sample_random_points(glacier_mask, sig_coords_arr, num_pts, radius, row_limit, W, min_dist=2*radius):
    """Spatial rejection sampling to get random control points inside glacier."""
    rand_coords = []
    attempts = 0
    while len(rand_coords) < num_pts and attempts < 20000:
        r = random.randint(radius, row_limit - radius - 1)
        c = random.randint(radius, W - radius - 1)
        if glacier_mask[r, c]:
            dist_to_sig = np.linalg.norm(sig_coords_arr - np.array([r, c]), axis=1)
            if np.all(dist_to_sig >= min_dist):
                if not rand_coords or np.all(np.linalg.norm(np.array(rand_coords) - np.array([r, c]), axis=1) >= min_dist):
                    rand_coords.append((r, c))
        attempts += 1
    return rand_coords

def compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius):
    sig_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in sig_coords]
    rand_fsim = [get_masked_fsim_metrics(before_img, after_img, glacier_mask, r, c, radius) for r,c in rand_coords]
    sig_fsim = [f for f in sig_fsim if f is not None]
    rand_fsim = [f for f in rand_fsim if f is not None]
    return np.mean(sig_fsim), np.mean(rand_fsim)

# ---------------------------
# 3. LOAD GLACIER MASK
# ---------------------------
# Load one image to get size & transform
with rasterio.open(date_pairs[0][1]) as ds:
    transform_ps = ds.transform
    crs_ps = ds.crs
    H, W = ds.height, ds.width

gdf = gpd.read_file(shapefile_path).to_crs(crs_ps)
glacier_mask = features.rasterize(
    shapes=gdf.geometry,
    out_shape=(H, W),
    transform=transform_ps,
    fill=0,
    all_touched=True,
    default_value=1
).astype(bool)

# Latitude → row conversion
row_limit = int((transform_ps.f - lat_cutoff) / abs(transform_ps.e))
row_limit = np.clip(row_limit, 0, H-1)

# ---------------------------
# 4. LOOP OVER DATES
# ---------------------------
dates = []
sig_avg_list = []
rand_avg_list = []

for label, before_file, after_file, sar_file in date_pairs:
    # Load before/after images
    with rasterio.open(before_file) as ds:
        before_img = ds.read(band_nir).astype(np.float32)
    with rasterio.open(after_file) as ds:
        after_img = ds.read(band_nir).astype(np.float32)

    # Map SAR points
    sig_coords = map_sar_points_to_planetscope(sar_file, glacier_mask, transform_ps, row_limit, W, radius)
    sig_coords_arr = np.array(sig_coords)

    # Sample random points
    rand_coords = sample_random_points(glacier_mask, sig_coords_arr, len(sig_coords), radius, row_limit, W)

    # Compute averages
    sig_avg, rand_avg = compute_avg_fsim(before_img, after_img, glacier_mask, sig_coords, rand_coords, radius)

    dates.append(label)
    sig_avg_list.append(sig_avg)
    rand_avg_list.append(rand_avg)

    print(f"{label} — Avg FSIM: Sig={sig_avg:.4f}, Rand={rand_avg:.4f}")

# ---------------------------
# 5. PLOT LINE CHART
# ---------------------------
plt.figure(figsize=(12,6))
plt.plot(dates, sig_avg_list, marker='o', label='Significant Points')
plt.plot(dates, rand_avg_list, marker='o', label='Random Points')
plt.xlabel("Date of SAR Acquisition")
plt.ylabel("Average FSIM (Higher = Less Change)")
plt.title("Glacier Feature Similarity Over Time")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
